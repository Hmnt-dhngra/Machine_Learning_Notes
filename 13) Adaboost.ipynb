Adaboost rests on the foundations of Decision Trees and Random Forest. 

- In a random forest, each time you make a tree you usually make full sized trees i.e. there is no predetermined maximum depth. 
- In a forest of trees made with Adaboost, the trees are usually just a node and 2 leaves. 

*A tree with just one node and 2 leaves is called a stump.*
So Adaboost is really a forest of stumps. 

Stumps are not good at making accurate classifications. 

- In random forest each tree has an equal vote on the final classification. 
- In adaboost, some stumps higher vote than others, like the larger stumps have more vote than the smaller stumps

Independence of trees
- In random forest, it doesn't matter which tree was made first
- In adaboost forest of stumps, the order of the trees is important. The errors that the first stump makes influences how the second stump is made.

##### **Main Ideas of AdaBoost**

- Adaboost combines a lot of weak learners to make classifications. The weak learners are almost always stumps. 
- Some stumps get more say in the classification than others. 
- Each stump is made by taking the previous stump's mistakes into account.
