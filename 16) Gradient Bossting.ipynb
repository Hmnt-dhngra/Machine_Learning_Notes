#### **Gradient Boosting**

##### **Theory**

##### **What is Gradient Boosting?**

Gradient Boosting is an ensemble learning technique that builds a strong predictive model by sequentially combining multiple weak learners, typically shallow decision trees called decision stump {also meaning depth of 1} . Small small decision trees with max depth of 1. 

It’s especially famous for powering models like XGBoost, LightGBM, and CatBoost.

Unlike bagging-based methods (e.g., Random Forest), where models are trained independently and in parallel, gradient boosting trains models sequentially, with each new model focusing on correcting the mistakes made by the previous ensemble.

*Hundreds of stumps creeate a boosting mechanism!*

##### **Working**
- Start with an initial model (often a simple constant prediction).
- Compute the errors (residuals) of the current model.
- Train a new weak learner to predict these residuals.
- Add this learner to the existing model with a scaling factor (learning rate).
- Repeat the process for a fixed number of iterations or until convergence.

*Each step “boosts” performance by correcting what came before.*



##### **Why is it called gradient boosting**

Because it uses the gradient of the loss function to decide:
- What the next model should learn
- Where the current model is most wrong

Mathematically, it’s doing gradient descent in function space:

- The “gradient” tells you the direction to reduce error.
- Each new tree approximates that gradient.

**Why is it needed?**

**Pruning :** Removing the splits that do not reduce the loss enough.
