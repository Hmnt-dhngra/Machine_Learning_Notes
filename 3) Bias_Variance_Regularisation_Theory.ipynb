## Bias and Variance

#### **Bias** 

The inability of a ML Model like Linear Regression to capture true relationships is called as bias. 

Bias is error caused by wrong assumptions in the ML algorithm.

*Intuition* : Bias is how far your model’s average prediction is from the true function.

When the relationship is not captured properly it is a high bias system

- High bias → model is too simple.  **Bad relationship captured = Higher bias**
- It cannot capture the true pattern in the data

- Leads to systematic error


**Higher Bias Symptoms:**
- Poor performance on training data
- Poor performance on test data
- This is called underfitting

#### **Variance**

The difference in fits of train and test is called variance. **Bigger the difference higher is variance**

Variance is error caused by sensitivity to the training data.

- High variance → model is too complex
- Small changes in data lead to large changes in the model
- The model learns noise instead of signal

*Intuition* : Variance measures how much your model would change if you trained it on a different sample of the same dataset.


**Higher Variance Symptoms :** 

- Excellent performance on training data
- Poor performance on test data
- This is called overfitting

#### **Is There a Bias–Variance Tradeoff?**
Yes—and this is unavoidable !!!

As model complexity increases:

- Bias decreases (model becomes more expressive)
- Variance increases (model becomes more sensitive)

As model complexity decreases:

- Bias increases
- Variance decreases

This tension is the bias–variance tradeoff.

You cannot minimize both simultaneously—improving one usually worsens the other.

In Ideal world, you need low bias and low variance, but it is usually hard to achieve, Hence the analogy of bias variance tradeoff. 

**REAL WORLD EXPECTATIONS FROM THE MODEL** 

High bias but Low variance because we want the model to show similar accuracy like what we achieved in training so that we are not given any suprises later in testing. 

From the bias–variance discussion, we learned:

- High variance models fit training data too well
- They learn noise, not just signal
- This hurts generalization

So the real problem is:

**How do we stop a model from becoming too sensitive to the training data without making it too simple?**

That question is regularisation.

**What is Regularisation?**

Regularisation is any technique that intentionally restricts a model’s flexibility in order to reduce variance and improve generalization.

In other words:

- We accept a bit more bias
- To significantly reduce variance
- And get lower test error

This is the bias–variance tradeoff, made explicit and controlled.

**“Fit the data well — but prefer simpler explanations.”**

**Regularisation** : 
- A technique that will help us achieve a model that works good for both training and test dataset. 
- *Working Fundamental*: Don't minimise the error, just keep the coefficients controlled and not overfit the train data.
- Regularization is a machine learning technique that prevents overfitting by penalizing model complexity. 
- It works by adding a penalty term to the loss function, which discourages the model from learning noise in the training data and helps it generalize better to new, unseen data. 

Regularisation are of 3 types :
- **L1 Regularisation** (Lasso Regression) : Used for feature selection i.e. features which are not very important won't be considered (handle multicollinearity) (sum of the absolute values of the coefficients). Shrinks features to nearly 0 which are less relevant. 
- **L2 Regularisation** (Ridge Regression) : Used for preventing overfitting. (sum of the squared coefficients). Shrinks features but never make them to be 0
- **Elastic Net Regression** : Combination of L1 & L2 hence reduces overfitting and helps with the feature selection.
